```python
import pandas as pd
df = pd.read_csv("Riched_Carnivorous_Plant.csv")
```

```python
display(df.info())
```

```python
print("Unique values in 'range' column:")
display(df['range'].value_counts())
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
from sklearn.preprocessing import StandardScaler

# Filter out 'unknown' values from the 'range' column
df_filtered = df[df['range'] != 'unknown'].copy()

# Encode the 'range' column to a binary numerical target variable
df_filtered['range_encoded'] = df_filtered['range'].apply(lambda x: 1 if x == 'inside' else 0)

# Define features (X) and target (y)
features = ['longitude', 'latitude', 'elevation'] + [f'bio{i}' for i in range(1, 20)]
X = df_filtered[features]
y = df_filtered['range_encoded']

# Initialize StandardScaler
scaler = StandardScaler()

# Scale the features
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=features, index=X.index)

print("First 5 rows of the processed data with target variable:")
display(df_filtered[['range', 'range_encoded']].head())

print("\nValue counts for the encoded target variable:")
display(y.value_counts())
```

```python
# Split data into training and testing sets (using scaled features)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Initialize and train the Logistic Regression model with class_weight='balanced'
model = LogisticRegression(max_iter=1000, solver='liblinear', class_weight='balanced')
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
print("\nLogistic Regression Model Performance (after scaling, increased max_iter/solver, and class_weight='balanced'):")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
