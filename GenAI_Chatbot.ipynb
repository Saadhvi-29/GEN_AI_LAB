```python
import os
from getpass import getpass

os.environ["OPENAI_API_KEY"] = getpass("Enter your OpenAI API key: ")
```

```python
import openai
import gradio as gr

openai.api_key = os.environ["OPENAI_API_KEY"]
```

```python
def chatbot_response(user_input):
    """
    Lab Patterns implemented:
    - Persona Prompt
    - Cognitive Verifier
    - Question Refinement
    - Provide New Info + Ask Questions
    - Root Prompt
    """

    prompt = f"""
You are an AI tutor (Persona Prompt).
Always answer clearly and correctly. Verify your answer before responding (Cognitive Verifier).
Refine the user's question if needed (Question Refinement).

User question: "{user_input}"

Instructions:
- Give a complete, accurate answer.
- Provide one new piece of information related to the topic.
- Ask a follow-up question at the end.
- Do not repeat the user's question.

Answer:
"""

    try:
        # ✅ New API (1.x+)
        response = openai.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": prompt}],
            temperature=0,
            max_tokens=250
        )

        answer = response.choices[0].message.content.strip()
        return answer

    except Exception as e:
        return f"Error: {str(e)}"
```

```python
chatbot = gr.Interface(
    fn=chatbot_response,
    inputs=gr.Textbox(lines=2, placeholder="Ask your question here..."),
    outputs="text",
    title="Lab Chatbot with OpenAI API (Updated 1.x)",
    description="Demonstrates all 5 prompt patterns using OpenAI GPT-3.5 with the new API"
)

chatbot.launch()
```
